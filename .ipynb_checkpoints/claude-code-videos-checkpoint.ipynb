```python
import csv
import re
import time
from datetime import datetime
from urllib.parse import urljoin

import requests
from bs4 import BeautifulSoup
```

```python
## this is for scraping ALL videos YouTube so you don't have to run this cell 
## you can find the output in Assignment 1 
## but also run it if you want idc 

BASE = "https://edworkforce.house.gov"
HEADERS = {"User-Agent": "Mozilla/5.0 (research scraper; contact: you@example.com)"}
START_DATE = datetime(2022, 11, 1)
YEARS = range(2022, datetime.now().year + 1)

YT_RE = re.compile(r"(youtube\.com|youtu\.be)", re.I)
DATE_RE = re.compile(r"([A-Z][a-z]+ \d{1,2}, \d{4})")

session = requests.Session()
session.headers.update(HEADERS)
```

```python
## more code pulling the YouTube link data (published date, hearing title, etc)
## don't have to re-run if you have the file already

def get_soup(url, retries=3):
    for attempt in range(retries):
        try:
            r = session.get(url, timeout=30)
            r.raise_for_status()
            return BeautifulSoup(r.text, "html.parser")
        except requests.RequestException as e:
            if attempt == retries - 1:
                raise
            time.sleep(2 * (attempt + 1))


def parse_list_page(year):
    """Return a list of dicts: title, date, subcommittee, event_url."""
    url = f"{BASE}/calendar/list.aspx?EventTypeID=189&Year={year}"
    soup = get_soup(url)
    events, seen = [], set()

    for a in soup.find_all("a", href=True):
        if "eventsingle.aspx" not in a["href"].lower():
            continue
        event_url = urljoin(url, a["href"])   # join to the list page URL, not BASE
        if event_url in seen:
            continue
        seen.add(event_url)

        title = a.get_text(" ", strip=True)

        # The surrounding table cell holds: title / subcommittee / date / subcommittee
        cell = a.find_parent("td") or a.parent
        lines = [l for l in cell.get_text("\n", strip=True).split("\n") if l]

        date_obj, subcommittee = None, ""
        for line in lines:
            m = DATE_RE.search(line)
            if m and not date_obj:
                date_obj = datetime.strptime(m.group(1), "%B %d, %Y")
        for line in lines:
            if line != title and not DATE_RE.search(line):
                subcommittee = line
                break

        events.append({
            "title": title,
            "date": date_obj,
            "subcommittee": subcommittee,
            "event_url": event_url,
        })
    return events

def get_youtube_links(event_url):
    soup = get_soup(event_url)
    links = []
    for a in soup.find_all("a", href=True):
        href = urljoin(event_url, a["href"])
        if YT_RE.search(href) and "/channel/" not in href and href not in links:
            links.append(href)
    return links

def main():
    all_events = []
    for year in YEARS:
        try:
            events = parse_list_page(year)
        except requests.RequestException as e:
            print(f"Could not load {year}: {e}")
            continue
        print(f"{year}: found {len(events)} events")
        all_events.extend(events)
        time.sleep(1)

    # keep only events on/after Nov 1, 2022 < I chose this date bc ChatGPT launched in Nov. 2022 
    all_events = [e for e in all_events if e["date"] is None or e["date"] >= START_DATE]
    all_events.sort(key=lambda e: e["date"] or datetime.min, reverse=True)

    with open("edworkforce_hearings.csv", "w", newline="", encoding="utf-8-sig") as f:
        w = csv.writer(f)
        w.writerow(["title", "date", "subcommittee", "youtube_link", "event_page"])
        for i, e in enumerate(all_events, 1):
            try:
                yt = get_youtube_links(e["event_url"])
            except requests.RequestException as err:
                print(f"  failed: {e['event_url']} ({err})")
                yt = []
            w.writerow([
                e["title"],
                e["date"].strftime("%Y-%m-%d") if e["date"] else "",
                e["subcommittee"],
                "; ".join(yt),
                e["event_url"],
            ])
            print(f"[{i}/{len(all_events)}] {e['title'][:60]} -> {len(yt)} link(s)")
            time.sleep(1)

    print("Saved edworkforce_hearing_links.csv")


if __name__ == "__main__":
    main()
```

```python
## pulling captions from the YouTube videos

INPUT_CSV = "assignment 1/data/ai_edworkforce_hearings.csv" ## this is where YOUR csv list with antisemitism hearings goes 
CAPTION_DIR = "captions"        # raw .vtt files
TRANSCRIPT_DIR = "transcripts"  # clean .txt files
INDEX_CSV = "transcript_index.csv"
INCLUDE_TIMESTAMPS = False      # True -> a [HH:MM:SS] marker before each line
DELAY_SECONDS = 2               # pause between videos

import random

DELAY_SECONDS = 8        # base pause between videos < I was getting 429 errors for scraping from YouTube oops! 
MAX_RETRIES = 5          # attempts per video on 429 errors

os.makedirs(CAPTION_DIR, exist_ok=True)
os.makedirs(TRANSCRIPT_DIR, exist_ok=True)

VIDEO_ID_RE = re.compile(r"(?:v=|youtu\.be/|/live/|/embed/|/shorts/)([A-Za-z0-9_-]{11})")
TIMESTAMP_RE = re.compile(r"^(\d{2}):(\d{2}):(\d{2})\.\d{3}\s+-->")
TAG_RE = re.compile(r"<[^>]+>")


# ---------------- import ----------------
def load_jobs(path):
    jobs = []
    with open(path, newline="", encoding="utf-8-sig") as f:
        for row in csv.DictReader(f):
            links = [l.strip() for l in row["youtube_link"].split(";") if l.strip()]
            for url in links:
                m = VIDEO_ID_RE.search(url)
                if not m:
                    print(f"Skipping (no video ID found): {url}")
                    continue
                jobs.append({
                    "title": row["title"],
                    "date": row["date"],
                    "subcommittee": row["subcommittee"],
                    "url": url,
                    "video_id": m.group(1),
                })
    return jobs


# ---------------- download captions ----------------
def download_captions(url):
    opts = {
        "skip_download": True,
        "writeautomaticsub": True,
        "subtitleslangs": ["en"],
        "subtitlesformat": "vtt",
        "outtmpl": f"{CAPTION_DIR}/%(id)s",
        "quiet": True,
        "no_warnings": True,
        "sleep_interval_subtitles": 3,   # yt-dlp's own pause before each subtitle request
        "retries": 3,
        "extractor_retries": 3,
    }
    for attempt in range(1, MAX_RETRIES + 1):
        try:
            with yt_dlp.YoutubeDL(opts) as ydl:
                ydl.download([url])
            return
        except Exception as e:
            if "429" in str(e) and attempt < MAX_RETRIES:
                wait = 30 * 2 ** (attempt - 1) + random.uniform(0, 10)   # ~30s, 60s, 120s, 240s
                print(f"  429 rate limit, waiting {wait:.0f}s (attempt {attempt}/{MAX_RETRIES})")
                time.sleep(wait)
            else:
                raise


def find_vtt(video_id):
    matches = glob.glob(f"{CAPTION_DIR}/{video_id}*.vtt")
    return matches[0] if matches else None


# ---------------- clean VTT ----------------
def clean_vtt(vtt_path, with_timestamps=False):
    """Convert a YouTube auto-caption .vtt into plain text.
    Auto-captions repeat each line across cues, so consecutive duplicates are dropped."""
    entries, last_text, current_start = [], None, 0

    with open(vtt_path, encoding="utf-8") as f:
        for raw in f:
            line = raw.strip()
            if not line or line.startswith(("WEBVTT", "Kind:", "Language:", "NOTE")):
                continue
            m = TIMESTAMP_RE.match(line)
            if m:
                h, mi, s = map(int, m.groups())
                current_start = h * 3600 + mi * 60 + s
                continue
            text = TAG_RE.sub("", line).strip()
            text = text.replace("&amp;", "&").replace("&gt;", ">").replace("&lt;", "<")
            if text and text != last_text:
                entries.append((current_start, text))
                last_text = text

    if with_timestamps:
        return "\n".join(
            f"[{s // 3600:02d}:{(s % 3600) // 60:02d}:{s % 60:02d}] {t}" for s, t in entries
        )
    return " ".join(t for _, t in entries)


# ---------------- export ----------------
def write_transcript(job, text):
    name = f"{job['date'] or 'undated'}_{job['video_id']}.txt"
    path = os.path.join(TRANSCRIPT_DIR, name)
    tmp = path + ".part"
    with open(tmp, "w", encoding="utf-8") as f:
        f.write(f"Title: {job['title']}\n")
        f.write(f"Date: {job['date']}\n")
        f.write(f"Subcommittee: {job['subcommittee']}\n")
        f.write(f"URL: {job['url']}\n\n")
        f.write(text)
    os.replace(tmp, path)
    return path


def transcript_path(job):
    return os.path.join(TRANSCRIPT_DIR, f"{job['date'] or 'undated'}_{job['video_id']}.txt")


# ---------------- main ----------------
def main():
    jobs = load_jobs(INPUT_CSV)
    print(f"{len(jobs)} videos to process")

    index_rows = []
    for i, job in enumerate(jobs, 1):
        out_path = transcript_path(job)
        status = "ok"

        if os.path.exists(out_path):
            status = "already done"
        else:
            try:
                if not find_vtt(job["video_id"]):
                    download_captions(job["url"])
                vtt = find_vtt(job["video_id"])
                if not vtt:
                    status = "no captions available"
                else:
                    text = clean_vtt(vtt, INCLUDE_TIMESTAMPS)
                    out_path = write_transcript(job, text)
            except Exception as e:
                status = f"error: {e}"
            time.sleep(DELAY_SECONDS+ random.uniform(0, 5))

        print(f"[{i}/{len(jobs)}] {job['video_id']} - {status}")
        index_rows.append({
            "title": job["title"], "date": job["date"],
            "subcommittee": job["subcommittee"], "url": job["url"],
            "video_id": job["video_id"],
            "transcript_file": out_path if status in ("ok", "already done") else "",
            "status": status,
        })

    with open(INDEX_CSV, "w", newline="", encoding="utf-8-sig") as f:
        w = csv.DictWriter(f, fieldnames=list(index_rows[0].keys()))
        w.writeheader()
        w.writerows(index_rows)
    print(f"Saved {INDEX_CSV}")


if __name__ == "__main__":
    main()
```
