```python
#borrowing this code from Leon 

import os
import glob
import json
import gzip

import pandas as pd
from ollama import chat
from pydantic import BaseModel
```

```python
## importing the files from the transcripts export folder 
ed_transcripts = "transcripts"

def load_transcript(path):
    with open(path, encoding="utf-8") as f:
        raw = f.read()

    # header and body are separated by the first blank line
    header, _, body = raw.partition("\n\n")

    meta = {}
    for line in header.splitlines():
        key, _, value = line.partition(": ")
        meta[key.strip().lower()] = value.strip()

    return {
        "title": meta.get("title", ""),
        "date": meta.get("date", ""),
        "subcommittee": meta.get("subcommittee", ""),
        "url": meta.get("url", ""),
        "video_id": os.path.splitext(os.path.basename(path))[0].split("_")[-1],
        "file": path,
        "text": body.strip(),
    }

def load_all(folder=ed_transcripts):
    paths = sorted(glob.glob(os.path.join(folder, "*.txt")))
    df = pd.DataFrame([load_transcript(p) for p in paths])
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df["word_count"] = df["text"].str.split().str.len()
    return df.sort_values("date").reset_index(drop=True)

df = load_all()
print(df[["date", "title", "subcommittee", "word_count"]])
print(f"\n{len(df)} transcripts loaded")
```

```python
##stealing this code from Leon

keywords_ai = [
    'artificial intelligence', 'AI', 'AI-assisted', 'large language models', 'GenAI', 'generative AI', 'generative artificial intelligence'
]
df['contains_keywords_ai'] = df['text'].str.contains(
    '|'.join(keywords_ai), regex=True, case=False
)
```

```python
## lifting this from Leon's code 
class AILabel(BaseModel):
    """
    This defines the data schema of the LLM response
    """
    contains_ai_llm: bool | None
    quote: str

dir_labels = "output/ai_labels"
os.makedirs(dir_labels, exist_ok=True)

for i, row in df.iterrows():
    # save one label per transcript segment as a record
    fn_out = dir_labels + f"{row['video_id']}-{row['id']}.json"
    
    if not os.path.exists(fn_out):
        prompt = f"Does this hearing transcript mention AI?\n\n{row.text}"          
        response = chat(
            model="qwen3.5:4b",
            messages=[{
                "role": "user", 
                "content": prompt
            }],
            format=AILabel.model_json_schema(),
            think=False,
        )
        
        label = AILabel.model_validate_json(
            response.message.content
        )
            
        with open(fn_out, "w") as f:
            f.write(label.model_dump_json())
```

```python
labels = []
for fn in glob.glob(dir_labels + "*.json"):
    video_id, segment_id = fn.split("/")[-1].split(".json")[0].rsplit("-", 1)
    with open(fn) as f:
        label = json.load(f)
    labels.append({"video_id": video_id, "id": int(segment_id), **label})

df_labels = pd.DataFrame(labels)
df_labels['contains_ai_llm'].value_counts()
```
